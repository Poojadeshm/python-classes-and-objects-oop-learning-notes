```python
class shop:                   # shop nam ka class bnaya hai
    def __init__(self):       #constructor
        self.items = []       #ek empty list banayi jisme items store honge

    def add_item(self, item):
        self.items.append(item) #appenend is used to add items in list
        print(f"{item} added")

    def show_items(self):
        print("Items:", self.items)


s1 = shop()               #shop instance
s1.add_item("Milk")
s1.add_item("Bread")
s1.show_items()          #this will show item list
```

```python
class student:                               #-----------creating a student template
    def __init__(self, name, marks):         #--------_init__ (Constructor)
        
 #store value or data   
        self.name = name                 #-saving values inside object
        self.marks = marks
        print(f"Marks: {self.marks}")
        
#display() function
    def display(self):                        
        print(f"Name: {self.name}")
        print(f"Marks: {self.marks}")
        

    def result(self):
        if self.marks > 40:
            print("Pass")
        else:
            print("Fail")


s1 = student("Pooja", 85)
s1.display()
s1.result()
```

```python
class pet:
    def __init__(self, name,age):
        self.name = name
        self.age =age
       

    def message(self):
        print(f"My name is {self.name}")
        print(f"My age is {self.age}")
    
    def sleep(self):
        print("sleeping")

    def wakeup(self):
        self.sleep()
        print("wakeup")    

class dog(pet):
    def __init__(self,name,age,breed):
        super().__init__(name,age)
        self.breed=breed

    def speak(self):
        print(f"I am a dog of breed {self.breed}")
        print("I bark") 

class cat(pet):

    def __init__(self,name,age,breed):
        super().__init__(name,age)
        self.breed=breed

    def speak(self):
        print("I am a Cat")
        print("I meaow") 
```

```python
d2=dog("ceaser",2,"husky")
d2.message()
d2.speak()
d2.wakeup()
```

```python
#class laptop
class laptop:
    def __init__(self, brand, ram):
        self.brand = brand
        self.ram = ram

    def check(self):
        if self.ram >= 8:
            print("Good performance")
        else:
            print("Low performance")


l1 = laptop("HP", 4)
l1.check()

l2 = laptop("Dell", 16)
l2.check()
```

```python
#EMPLOYEE Class
class employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def work(self):
        print(f"{self.name} is working")

    def bonus(self):
        print(f"Bonus: {self.salary * 0.10}")


e1 = employee("Rahul", 50000)
e1.work()
e1.bonus()
```

```python
#ATM Class
class atm:
    def __init__(self, balance):
        self.balance = balance

    def check_balance(self):
        print(f"Balance: {self.balance}")

    def deposit(self, amount):
        self.balance += amount
        print("Money Deposited")

    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdraw Successful")
        else:
            print("Insufficient Balance")


a1 = atm(1000)
a1.check_balance()
a1.deposit(500)
a1.withdraw(200)
a1.check_balance()
```

```python
class teacher:
    def __init__(self, name, subject):
        self.name = name
        self.subject = subject

    def display(self):
        print(f"Teacher: {self.name}")
        print(f"Subject: {self.subject}")

    def teach(self):
        print(f"{self.name} is teaching {self.subject}")


t1 = teacher("Sharma Sir", "Math")
t1.display()
t1.teach()
```

```python
class mobile:
    def __init__(self, brand, price):
        self.brand = brand
        self.price = price

    def call(self):
        print(f"{self.brand} is calling...")

    def show_price(self):
        print(f"Price is {self.price}")


m1 = mobile("Samsung", 20000)
m1.call()
m1.show_price()
```

```python
#Inheritance Example (Vehicle → Car, Bike)

class vehicle:
    def __init__(self, name):
        self.name = name

    def start(self):
        print(f"{self.name} is starting")


class car(vehicle):
    def speak(self):
        print("Car is running on 4 wheels")


class bike(vehicle):
    def speak(self):
        print("Bike is running on 2 wheels")


c = car("BMW")
c.start()
c.speak()

b = bike("Yamaha")
b.start()
b.speak()
```
